```python
def chatbot():
    print("--------------------------------------------------")
    print("Bot: Hello! I am your AI assistant. Type 'exit' or 'bye' to end our chat.")
    print("--------------------------------------------------")

    while True:
        # Get user input and convert to lowercase to handle variations in typing
        user_input = input("You: ").strip().lower()

        # Check for exit commands first
        if user_input in ["exit", "bye", "goodbye", "quit"]:
            print("Bot: It was lovely chatting with you! Have a wonderful day ahead. Goodbye!")
            break

        # Greeting logic
        elif any(greeting in user_input for greeting in ["hi", "hello", "hey"]):
            print("Bot: Hello there! It's great to talk to you. How can I help you today?")

        # Name / Introduction logic
        elif "your name" in user_input or "who are you" in user_input:
            print("Bot: My name is Aura! I'm a simple rule-based virtual assistant here to chat with you.")

        # How's your day / How are you logic
        elif "how is your day" in user_input or "how are you" in user_input:
            print("Bot: I'm doing wonderfully, thank you for asking! I hope you're having a great day as well.")

        # Gratitude logic
        elif "thank" in user_input or "thanks" in user_input:
            print("Bot: You're very welcome! It's always my pleasure to help.")

        # Fallback response for unrecognized inputs
        else:
            print("Bot: I hear you! I'm still learning, so I might not understand everything yet, but I'm here to listen.")

# Run the chatbot
if __name__ == "__main__":
    chatbot()
```
