```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Define Dataset
data = {
    'item_id': [1, 2, 3, 4, 5, 6],
    'title': [
        'Python for Data Science',
        'Web Development Bootcamp',
        'Machine Learning Essentials',
        'UI/UX Design Basics',
        'Deep Learning & AI Specialization',
        'Frontend JavaScript Guide'
    ],
    'description': [
        'Learn Python programming, pandas, data analysis, and visualization tools.',
        'Full stack web development using HTML, CSS, JavaScript, and Node js.',
        'Supervised learning, classification, regression, and scikit-learn models.',
        'User interface design, Figma, user experience research, and prototyping.',
        'Neural networks, deep learning, PyTorch, computer vision, and AI.',
        'Modern JavaScript, React, web design, and frontend development principles.'
    ]
}

df = pd.DataFrame(data)

# 2. Recommendation Logic Function
def recommend_items(user_input, top_n=3):
    all_texts = df['description'].tolist() + [user_input]
    
    vectorizer = TfidfVectorizer(stop_words='english')
    tfidf_matrix = vectorizer.fit_transform(all_texts)
    
    item_vectors = tfidf_matrix[:-1]
    user_vector = tfidf_matrix[-1]
    
    similarity_scores = cosine_similarity(user_vector, item_vectors).flatten()
    
    results = df.copy()
    results['similarity_score'] = similarity_scores
    
    return results.sort_values(by='similarity_score', ascending=False).head(top_n)

# 3. Take User Input and Display Output
user_interest = input("Enter your interest (e.g., 'python and data analysis'): ")

recommendations = recommend_items(user_interest, top_n=3)

print("\n--- Recommended Items ---")
for idx, row in recommendations.iterrows():
    print(f"Title: {row['title']}")
    print(f"Match Score: {round(row['similarity_score'] * 100, 2)}%\n")
```
